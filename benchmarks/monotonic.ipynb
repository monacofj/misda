# MISDA monotonic discrimination benchmark

Controlled cases designed to distinguish linear Pearson dependence from monotonic Spearman dependence while retaining linear, non-monotonic, and independent controls.

```python
from benchmarks.run_monotonic import run_monotonic

N = 300
SEED = 123
artifact = run_monotonic(n=N, seed=SEED)
```

```python
import pandas as pd

pd.DataFrame([
    {
        'case': case['id'],
        'pearson': case['focus_correlation']['pearson'],
        'spearman': case['focus_correlation']['spearman'],
        'dl': case['latent_observed'],
        'dl_expected': case['latent_expected'],
        'ds': case['structural_observed'],
        'ds_expected': case['structural_expected'],
        'exact': case['exact'],
    }
    for case in artifact['cases']
])
```

```python
for case in artifact['cases']:
    print(f"\n{case['id']}")
    print(f"  {case['description']}")
    print(f"  G+  : {case['g_positive_edges']}")
    print(f"  G±  : {case['g_signed_edges']}")
    print(f"  MIS : {case['mis_count']}; selected={case['selected_objectives']}")
    print(f"  separation={case['separation_status']}; support={case['support_status']}")
```
